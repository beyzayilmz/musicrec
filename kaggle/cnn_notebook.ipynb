## Adım C: ResNet-34 (sıfırdan)
Aynı log-mel verisi, `--arch resnet34` (~21 M parametre). Düzenlileştirme: mixup, ağırlık EMA'sı, stochastic depth, rastgele kazanç, SpecAugment, pos-weight. Sıra: duman testi -> tek seed tam koşu (T4'te epoch süresini not et) -> 3-5 seed. Overfit görürsen (train loss düşüyor, val PR-AUC erken tepe yapıp düşüyor) önce `--width 48` veya `--arch resnet18` dene.

```python
!python /kaggle/temp/musicrec/src/train_cnn.py --emb-dir $EMB --out-dir /kaggle/working/runs_resnet \
    --arch resnet34 --epochs 2 --max-train 500 --seeds 1 --pos-weight --ema 0.999
```

```python
RESNET = ("--arch resnet34 --epochs 80 --patience 15 --batch-size 64 --lr 1e-3 --weight-decay 5e-2 "
          "--crop-sec 15 --mixup 0.3 --ema 0.999 --drop-path 0.1 --gain-db 6 --clip 5 --pos-weight")
!python /kaggle/temp/musicrec/src/train_cnn.py --emb-dir $EMB --out-dir /kaggle/working/runs_resnet $RESNET --seeds 1
```
